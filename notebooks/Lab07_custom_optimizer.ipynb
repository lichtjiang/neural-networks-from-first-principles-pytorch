```python
import torch
from torch import nn
```

```python
# Custom optimizer
def UpdateWeights(parameters, learning_rate=0.1):
    with torch.no_grad():
        for parameter in parameters:
            parameter -= learning_rate * parameter.grad
```

```python
class TinyNet(nn.Module):

    def __init__(self):
        super().__init__()

        self.w1 = nn.Parameter(torch.tensor([[0.5]]))
        self.b1 = nn.Parameter(torch.tensor([0.0]))

        self.w2 = nn.Parameter(torch.tensor([[-0.5]]))
        self.b2 = nn.Parameter(torch.tensor([0.0]))

        self.v1 = nn.Parameter(torch.tensor([[1.0]]))
        self.v2 = nn.Parameter(torch.tensor([[1.0]]))

        self.b3 = nn.Parameter(torch.tensor([0.0]))

    def forward(self, x):

        z1 = x * self.w1 + self.b1
        a1 = torch.tanh(z1)

        z2 = x * self.w2 + self.b2
        a2 = torch.tanh(z2)

        y_hat = a1 * self.v1 + a2 * self.v2 + self.b3

        return y_hat
```

```python
model = TinyNet()

x = torch.tensor([[1.0]])
y = torch.tensor([[1.0]])

y_hat = model(x)

print(f"y_hat: {y_hat}\n")

delta = y_hat - y

print(f"y_hat - y:  {delta}\n")

loss = delta.pow(2)

print(f"loss: {loss}\n")

print(loss.is_leaf)         # False — it's the result of operations
```

```python
loss.backward()
```

```python
for name, parameter in model.named_parameters():
    print(f"name: {name}")
    print(parameter)
    print(f"parameter grad: {parameter.grad}")
    print()
```

```python
UpdateWeights(model.parameters(), 0.1)
```
